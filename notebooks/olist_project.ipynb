```python
import pandas as pd
import numpy as np
import os
```

```python

data_path = r"C:\Users\Rohit\Downloads\olist_dataset.csv" 

# Load all 9 CSV files
orders = pd.read_csv(os.path.join(data_path, "olist_orders_dataset.csv"))
customers = pd.read_csv(os.path.join(data_path, "olist_customers_dataset.csv"))
items = pd.read_csv(os.path.join(data_path, "olist_order_items_dataset.csv"))
payments = pd.read_csv(os.path.join(data_path, "olist_order_payments_dataset.csv"))
products = pd.read_csv(os.path.join(data_path, "olist_products_dataset.csv"))
sellers = pd.read_csv(os.path.join(data_path, "olist_sellers_dataset.csv"))
reviews = pd.read_csv(os.path.join(data_path, "olist_order_reviews_dataset.csv"))
geolocation = pd.read_csv(os.path.join(data_path, "olist_geolocation_dataset.csv"))
category_translation = pd.read_csv(os.path.join(data_path, "product_category_name_translation.csv"))

# Check dataset dimensions
datasets = {
    "Orders": orders,
    "Customers": customers,
    "Items": items,
    "Payments": payments,
    "Products": products,
    "Sellers": sellers,
    "Reviews": reviews,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}

for name, df in datasets.items():
    print(name, ":", df.shape)
```

```python
# Inspect the structure of all 9 dataset
for name, df in datasets.items():

    print("\n" + "=" * 60)
    print("DATASET:", name)
    print("=" * 60)

    # Display column names
    print("\nColumns:")
    print(df.columns.tolist())

    # Display data types and non-null counts
    print("\nData Information:")
    df.info()
```

```python
for name, df in datasets.items():

    print("\nDataset:", name)

    missing = df.isnull().sum()

    # Display only columns containing missing values
    print(missing[missing > 0])
```

```python
for name, df in datasets.items():

    duplicate_count = df.duplicated().sum()

    print(name, ":", duplicate_count, "duplicate rows")
```

```python
# inspection report
inspection_report = []

for name, df in datasets.items():

    inspection_report.append({
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum()
    })

# Convert report into DataFrame
inspection_df = pd.DataFrame(inspection_report)

# Display report
display(inspection_df)
```

```python
# unique IDs
print("Orders:", orders["order_id"].nunique())
print("Customers:", customers["customer_id"].nunique())
print("Products:", products["product_id"].nunique())
print("Sellers:", sellers["seller_id"].nunique())
print("Reviews:", reviews["review_id"].nunique())

print("\nTotal Orders:", len(orders))
print("Unique Orders:", orders["order_id"].nunique())
```

```python
# Order date columns
order_date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
# Convert order dates
for col in order_date_columns:
    orders[col] = pd.to_datetime(
        orders[col],
        errors="coerce"
    )

# Convert review dates
review_date_columns = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in review_date_columns:
    reviews[col] = pd.to_datetime(
        reviews[col],
        errors="coerce"
    )

# Convert shipping limit date
items["shipping_limit_date"] = pd.to_datetime(
    items["shipping_limit_date"],
    errors="coerce"
)

# Check data types
print(orders[order_date_columns].dtypes)
print(reviews[review_date_columns].dtypes)
```

```python
# Missing value report
for name, df in datasets.items():

    print("\n", name.upper())

    missing_report = pd.DataFrame({
        "Missing_Count": df.isnull().sum(),
        "Missing_Percentage": (
            df.isnull().mean() * 100
        ).round(2)
    })

    print(missing_report[missing_report["Missing_Count"] > 0])
```

```python
# Create clean copies
orders_clean = orders.copy()
customers_clean = customers.copy()
items_clean = items.copy()
payments_clean = payments.copy()
products_clean = products.copy()
sellers_clean = sellers.copy()
reviews_clean = reviews.copy()
geolocation_clean = geolocation.copy()
category_translation_clean = category_translation.copy()

# Keep original missing review comments as NaN.
# Create separate analysis-friendly text columns.

reviews_clean["review_comment_title_analysis"] = (
    reviews_clean["review_comment_title"].fillna("No comment")
)

reviews_clean["review_comment_message_analysis"] = (
    reviews_clean["review_comment_message"].fillna("No comment")
)

# Keep product attributes missing where they are genuinely unknown.
# Do not invent dimensions or product weights.

print("Missing value handling completed.")
```

```python
# Count duplicates before cleaning
before = len(geolocation_clean)

print("Rows before:", before)

# Remove exact duplicate rows
geolocation_clean = geolocation_clean.drop_duplicates().copy()

# Reset index
geolocation_clean.reset_index(drop=True, inplace=True)

# Count after cleaning
after = len(geolocation_clean)

print("Rows after:", after)
print("Duplicates removed:", before - after)
```

```python
# Merge product data with English category translation
products_clean = products_clean.merge(
    category_translation_clean,
    on="product_category_name",
    how="left"
)
# Rename translated category column
products_clean.rename(
    columns={
        "product_category_name_english": "product_category_english"
    },
    inplace=True
)
# Check translated categories
print(
    products_clean[
        ["product_category_name", "product_category_english"]
    ].head(10)
)
# Check categories without translation
print(
    "Untranslated categories:",
    products_clean["product_category_english"].isnull().sum()
)
```

```python
products_clean["product_category_english"] = (
    products_clean["product_category_english"]
    .fillna("Unknown")
)
```

```python
# Calculate actual delivery duration in days
orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"] -
    orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / 86400

# Round to 2 decimal places
orders_clean["delivery_days"] = (
    orders_clean["delivery_days"].round(2)
)
# Display result
print(
    orders_clean[
        ["order_id", "delivery_days"]
    ].head(10)
)
```

```python
# Calculate difference between actual and estimated delivery
orders_clean["delay_days"] = (
    orders_clean["order_delivered_customer_date"] -
    orders_clean["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

orders_clean["delay_days"] = orders_clean["delay_days"].round(2)

# Positive = late
# Negative = delivered early
# Zero = delivered on estimated date
print(
    orders_clean[
        ["delay_days"]
    ].describe()
)
```

```python
# Identify orders delivered after the estimated date
orders_clean["is_late"] = np.where(
    orders_clean["order_delivered_customer_date"].notna() &
    orders_clean["order_estimated_delivery_date"].notna(),
    (
        orders_clean["order_delivered_customer_date"] >
        orders_clean["order_estimated_delivery_date"]
    ).astype(int),
    np.nan
)
# Check late delivery counts
print(orders_clean["is_late"].value_counts(dropna=False))
```

```python
# Extract year
orders_clean["order_year"] = (
    orders_clean["order_purchase_timestamp"].dt.year
)

# Extract month number
orders_clean["order_month_number"] = (
    orders_clean["order_purchase_timestamp"].dt.month
)

# Extract month name
orders_clean["order_month"] = (
    orders_clean["order_purchase_timestamp"].dt.strftime("%B")
)

# Create year-month column
orders_clean["year_month"] = (
    orders_clean["order_purchase_timestamp"].dt.to_period("M").astype("string")
)

# Display
print(
    orders_clean[
        ["order_purchase_timestamp", "order_year",
         "order_month", "year_month"]
    ].head(10)
)
```

```python
# Create item-level total
items_clean["item_total"] = (
    items_clean["price"] + items_clean["freight_value"]
)

# Aggregate item information at order level
order_value = items_clean.groupby("order_id").agg(
    product_revenue=("price", "sum"),
    freight_revenue=("freight_value", "sum"),
    order_value=("item_total", "sum"),
    total_items=("order_item_id", "count")
).reset_index()

# Merge order-level metrics
orders_clean = orders_clean.merge(
    order_value,
    on="order_id",
    how="left",
    validate="one_to_one"
)

# Display
print(
    orders_clean[
        ["order_id", "product_revenue",
         "freight_revenue", "order_value", "total_items"]
    ].head(10)
)
```

```python
# Function to identify outliers
def identify_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]
    print("\nColumn:", column)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Outlier count:", len(outliers))

    return outliers

# Check price outliers
price_outliers = identify_outliers(
    items_clean, "price"
)

# Check freight outliers
freight_outliers = identify_outliers(
    items_clean, "freight_value"
)
```

```python
# Convert ZIP prefixes into strings

customers_clean["customer_zip_code_prefix"] = (
    customers_clean["customer_zip_code_prefix"].astype("string").str.zfill(5)
)

sellers_clean["seller_zip_code_prefix"] = (
    sellers_clean["seller_zip_code_prefix"].astype("string").str.zfill(5)
)

geolocation_clean["geolocation_zip_code_prefix"] = (
    geolocation_clean["geolocation_zip_code_prefix"].astype("string").str.zfill(5)
)

# Standardize state codes
customers_clean["customer_state"] = (
    customers_clean["customer_state"].str.upper()
)

sellers_clean["seller_state"] = (
    sellers_clean["seller_state"].str.upper()
)

geolocation_clean["geolocation_state"] = (
    geolocation_clean["geolocation_state"].str.upper()
)

print("Geographical columns standardized.")
```

```python
# Check foreign key relationships
print("Items with missing order IDs:",
      (~items_clean["order_id"].isin(orders_clean["order_id"])).sum())

print("Items with missing product IDs:",
      (~items_clean["product_id"].isin(products_clean["product_id"])).sum())

print("Items with missing seller IDs:",
      (~items_clean["seller_id"].isin(sellers_clean["seller_id"])).sum())

print("Orders with missing customer IDs:",
      (~orders_clean["customer_id"].isin(customers_clean["customer_id"])).sum())

print("Payments with missing order IDs:",
      (~payments_clean["order_id"].isin(orders_clean["order_id"])).sum())

print("Reviews with missing order IDs:",
      (~reviews_clean["order_id"].isin(orders_clean["order_id"])).sum())
```

```python
# Store cleaned datasets
cleaned_datasets = {
    "orders": orders_clean,
    "customers": customers_clean,
    "items": items_clean,
    "payments": payments_clean,
    "products": products_clean,
    "sellers": sellers_clean,
    "reviews": reviews_clean,
    "geolocation": geolocation_clean,
    "category_translation": category_translation_clean
}

# Generate final cleaning report

final_report = []

for name, df in cleaned_datasets.items():

    final_report.append({
        "Dataset": name,
        "Rows": len(df),
        "Columns": len(df.columns),
        "Missing_Values": df.isnull().sum().sum(),
        "Duplicate_Rows": df.duplicated().sum()
    })

final_report_df = pd.DataFrame(final_report)

display(final_report_df)
```

```python
# Export all cleaned datasets

for name, df in cleaned_datasets.items():

    file_path = os.path.join(
        clean_path,
        name + "_cleaned.csv"
    )

    df.to_csv(file_path, index=False)

    print(name, "exported successfully!")

print("\nAll 9 cleaned datasets exported.")
```

```python

# Import required libraries
import pandas as pd
from sqlalchemy import create_engine, text, URL
from getpass import getpass

# MySQL connection details
username = "root"
password = getpass("Enter MySQL password: ")

host = "localhost"
port = 3306
database = "ecommerce_analytics"

# Create SQLAlchemy connection URL
connection_url = URL.create(
    drivername="mysql+pymysql",
    username=username,
    password=password,
    host=host,
    port=port,
    database=database
)

# Create database engine
engine = create_engine(
    connection_url,
    pool_pre_ping=True
)

print("MySQL engine created successfully!")
```

```python
# Test MySQL connection
try:
    with engine.connect() as connection:

        result = connection.execute(
            text("SELECT VERSION()")
        )
        version = result.fetchone()[0]

        print("MySQL connected successfully!")
        print("MySQL Version:", version)

except Exception as e:
    print("Connection failed!")
    print("Error:", e)
```

```python

clean_path = r"C:\Users\Rohit\Downloads\olist_dataset.csv\olist_cleaned.csv"

file_tables = {
    "orders_cleaned.csv": "orders",
    "customers_cleaned.csv": "customers",
    "items_cleaned.csv": "order_items",
    "payments_cleaned.csv": "payments",
    "products_cleaned.csv": "products",
    "sellers_cleaned.csv": "sellers",
    "reviews_cleaned.csv": "reviews",
    "geolocation_cleaned.csv": "geolocation",
    "category_translation_cleaned.csv": "category_translation"
}

for file_name, table_name in file_tables.items():

    # Build the complete CSV file path
    file_path = os.path.join(clean_path, file_name)

    # Check whether the file exists
    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        continue

    # Load the actual CSV file into Pandas
    df = pd.read_csv(file_path)

    print(f"\nUploading {table_name}...")
    print("Rows:", len(df))

    # Upload data to MySQL
    df.to_sql(
        name=table_name,
        con=engine,
        if_exists="fail",
        index=False,
        chunksize=5000
    )
    print(f"{table_name} uploaded successfully!")

print("\nUpload process completed!")
```
